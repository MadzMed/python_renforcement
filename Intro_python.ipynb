**Appeler une fonction**:


````python
def sum_of_square(a: int, b: int):
    square_a = a**2
    square_b = b**2
    return square_a + square_b
````

````python
sum_of_square(2, 3)
````

````python
def sum_of_square(*args, **kwargs):
    total = 0
    power = 2
    if kwargs.get("power"):
        power = kwargs["power"]
        
    for arg in args:
        total += arg ** power
    return total
````

````python
assert sum_of_square(2, 3, 4, 5) == 54 
assert sum_of_square(2, 3, 4, 5, power=3) == 224
````

````python
sum_of_square(2, 3, 4, 5, power=3)
````

args = **liste d'argument** que l'on peux donner à la fonction

````python
import numpy as np

def average(*args, **kwargs):
    total = 0
    count = np.sum(kwargs["weight"])

    print(kwargs)
    for index, arg in enumerate(args):
        total += arg * kwargs["weight"][index]

    return total / count
````

````python
average(1, 2, 50, weight=[5, 1, 1])
````

kwargs = **key words argument** que l'on récupère dans un dictionaire

````python
class Character: # CamelCase
    # caractéristique de l'objet qui va être crée )
    def __init__(self, name, surname, age) : # self => instance elle même
        self.name = name
        self.surname = surname
        self.age = age
        self.pv = 100
        self.level = 1

    # comportement: methode d'instance
    def battle(self):
        self.pv -= 10
        self._increment_level(1)

    def rest(self):
        self.pv = 100

    # priver
    def _increment_level(self, nb):
        self.level += nb


    
````

````python
luffy = Character("luffy", "Mugiwara", 17)
````

#### 4 principes:

##### Encapsulation

Avoir les caractéristiques ( ou données ) et le comportement ( les methodes ) qui sont regroupé dans la même Class !

trois types de methodes: private, public, interne

##### Abstraction

Masquer toute la compléxité du code ! fonctionnalités qui sont utiles ! 

##### Héritage

Possibilité de créer une classe à partir d'une autre classe ! elle a récupérer l'ensemble des caractéristique et comportement de la classe mère ! 

````python
class Animal:
    def __init__(self, name):
        self.name = name
        self.exist = True
        self.nb_of_pattoune = 8

    def make_noise(self):
        return "BRRRRRRR"

class Bird(Animal): # tu hérite de la classe animal
    def __init__(self, name):
        super().__init__(name) # instancie les info que la class animal a besoin 
        self.can_fly = True
````

````python
Bird("coucou").exist
````

##### Polymorphisme

des classes peuvent avoir le meme nom de method; mais le comportement est différent.

````python
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
        self.nb_of_pattoune = 4

    def make_noise(self):
        return "Woaf Woaf"
````

````python
class Cat(Animal):
    def __init__(self, name):
        super().__init__(name)

    def make_noise(self):
        return "Miaou Miaou"
````

````python
Dog("rex").nb_of_pattoune
````

````python
Cat("alibert").nb_of_pattoune
````

````python
Bird("coucou").make_noise()
````

````python
class Penguin(Bird):
    def __init__(self, name):
        super().__init__(name)

# le fait qu'il est la capacité de voler => héritage
````

````python
Penguin("pengui").can_fly
````

````python
class Manchot(Bird):
    def __init__(self, name):
        super().__init__(name)
        self.can_fly = False
# modifier son comportement => polymorphisme
````

````python
Manchot("manchot").can_fly
````
